```python
import pandas as pd
from google_play_scraper import reviews, Sort
```

# **Scraping DANA reviews from Google Play**

```python
# Scraping 20000 reviews of Dana from Google Play
all_reviews = []

target_jumlah_reviews = 20000
continuation_token = None

while len(all_reviews) < target_jumlah_reviews:
  result, continuation_token = reviews(
      'id.dana',
      lang='id',
      country='id',
      count=1000,
      sort=Sort.MOST_RELEVANT,
      continuation_token=continuation_token
  )
  all_reviews.extend(result)
```

```python
# Convert all_reviews list to a dataframe
df_reviews = pd.DataFrame(all_reviews)
```

```python
display(df_reviews.head())
```

# **Save the reviews to a CSV file**

```python
# Save the dataframe to a CSV file
df_reviews['content'].to_csv('dana_reviews.csv', index=False)

print("Reviews succesfully saved to 'dana_reviews.csv'")
```
